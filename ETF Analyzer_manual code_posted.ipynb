```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import date, datetime, timedelta

pd.set_option('display.max_columns', None)

# Load the dataset
def open_file(file_name: str) -> pd.DataFrame:
    return pd.read_csv(file_name, keep_default_na=False, na_values=[''])


# This creates an ETF column with one label per row.
def assign_etf_label(df, etf_name: str):
    return [etf_name] * len(df)


# This creates an Input_Date column with today's date for every row.
def assign_loading_date(df, input_date=None):
    if input_date is None:
        input_date = date.today()
    return [input_date] * len(df)


# The next block of code loads all holdings from relevant iShares ETFs offered by BlackRock

df_1 = open_file('AQLT.csv')
df_1['ETF'] = assign_etf_label(df_1, 'AQLT')
df_1['Input_Date'] = assign_loading_date(df_1)

df_2 = open_file('IEFA.csv')
df_2['ETF'] = assign_etf_label(df_2, 'IEFA')
df_2['Input_Date'] = assign_loading_date(df_2)

df_3 = open_file('IJH.csv')
df_3['ETF'] = assign_etf_label(df_3, 'IJH')
df_3['Input_Date'] = assign_loading_date(df_3)

df_4 = open_file('IVV.csv')
df_4['ETF'] = assign_etf_label(df_4, 'IVV')
df_4['Input_Date'] = assign_loading_date(df_4)

df_5 = open_file('IWY.csv')
df_5['ETF'] = assign_etf_label(df_5, 'IWY')
df_5['Input_Date'] = assign_loading_date(df_5)

df_6 = open_file('JPXN.csv')
df_6['ETF'] = assign_etf_label(df_6, 'JPXN')
df_6['Input_Date'] = assign_loading_date(df_6)

df_7 = open_file('EFAV.csv')
df_7['ETF'] = assign_etf_label(df_7, 'EFAV')
df_7['Input_Date'] = assign_loading_date(df_7)

df_8 = open_file('QUAL.csv')
df_8['ETF'] = assign_etf_label(df_8, 'QUAL')
df_8['Input_Date'] = assign_loading_date(df_8)

df_iShares = pd.concat([df_1, df_2, df_3, df_4, df_5, df_6, df_7, df_8], ignore_index=True)
df_iShares.head()
```

```python
# The next block of code is for data cleaning

# Test column *names*, not column *values* 

has_weight = 'Weight (%)' in df_iShares.columns
has_market = 'Market Weight' in df_iShares.columns

if has_weight or has_market:
    a = (
        pd.to_numeric(df_iShares['Weight (%)'], errors='coerce')
        if has_weight
        else pd.Series(np.nan, index=df_iShares.index)
    )
    b = (
        pd.to_numeric(df_iShares['Market Weight'], errors='coerce')
        if has_market
        else pd.Series(np.nan, index=df_iShares.index)
    )
    df_iShares['percent'] = a.combine_first(b)  # prefer Weight (%) when both exist


df_iShares.info()
```

```python
# The following code to ingest Vanguard and VanEck etfs

df_Vanguard = open_file('VCN.csv')
df_Vanguard['ETF'] = assign_etf_label(df_Vanguard, 'VCN')
df_Vanguard['Input_Date'] = assign_loading_date(df_Vanguard)

df_SMH = open_file('SMH.csv')
df_SMH['ETF'] = assign_etf_label(df_SMH, 'SMH')
df_SMH['Input_Date'] = assign_loading_date(df_SMH)

df_VanSMH = pd.concat([df_Vanguard, df_SMH], ignore_index=True)
df_VanSMH.head()

df_VanSMH['% of market value'] = pd.to_numeric(
    df_VanSMH['% of market value'].astype(str).str.replace('%', '', regex=False).str.strip(),
    errors='coerce'
)

df_VanSMH['% of market value% of Net Assets'] = pd.to_numeric(
    df_VanSMH['% of Net Assets'].astype(str).str.replace('%', '', regex=False).str.strip(),
    errors='coerce'
)

a = pd.to_numeric(df_VanSMH['% of market value'], errors="coerce")
b = pd.to_numeric(df_VanSMH['% of market value% of Net Assets'], errors="coerce")
df_VanSMH['percent'] = a.combine_first(b) # Weight (%) first
```

```python
# The following code to ingest Fidelity etfs

df_ftec = open_file('FTEC.csv')
df_ftec['ETF'] = assign_etf_label(df_ftec, 'FTEC')
df_ftec['Input_Date'] = assign_loading_date(df_ftec)

df_fqal = open_file('FQAL.csv')
df_fqal['ETF'] = assign_etf_label(df_fqal, 'FQAL')
df_fqal['Input_Date'] = assign_loading_date(df_fqal)

df_Fidel = pd.concat([df_ftec, df_fqal], ignore_index=True)
df_Fidel.head()
```

```python
#Data cleaning step for Fidelity ETFS

df_Fidel['percent'] = df_Fidel['Weight']
df_Fidel['Ticker'] = df_Fidel['Symbol']
df_Fidel.head()
```

```python
df_Fidel.info()
```

```python
df_Combined = pd.concat([df_iShares, df_VanSMH,df_Fidel], ignore_index=True)
```

```python
def ask_hold(prompt):
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Please enter a number.")
```

```python
holds = {}
for etf in df_Combined["ETF"].dropna().unique():
    holds[etf] = ask_hold(f"Enter dollar position for this ETF {etf!r}: ")

#If you want a column that repeats the dollar position on every row
df_Combined["my_positions"] = df_Combined["ETF"].map(holds)
df_Combined['my_values'] = df_Combined['my_positions'] * df_Combined['percent'] / 100


# this will be a future function
#my_hold = float(input("Enter your hold of this ETF: "))
#print(f"You entered: {my_hold}")
#df_iShares['my_value'] = df_iShares['percent'] * my_hold

```
